```python
# Weather Augmentation Module
import albumentations as A

def get_weather_transform():
    transform = A.Compose([
        A.RandomRain(p=0.3),
        A.RandomFog(p=0.3),
        A.RandomSnow(p=0.2),
        A.MotionBlur(p=0.3),
        A.RandomBrightnessContrast(p=0.4),
        A.GaussNoise(p=0.2)
    ])
    return transform
```

```python
# Image Enhancement Module
import cv2
import numpy as np

def clahe_enhancement(image):
    lab = cv2.cvtColor(image, cv2.COLOR_RGB2LAB)
    l, a, b = cv2.split(lab)

    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    cl = clahe.apply(l)

    merged = cv2.merge((cl, a, b))
    enhanced = cv2.cvtColor(merged, cv2.COLOR_LAB2RGB)

    return enhanced

def sharpen_image(image):
    kernel = np.array([[0,-1,0],
                       [-1,5,-1],
                       [0,-1,0]])
    return cv2.filter2D(image, -1, kernel)
```

```python
# Custom Dataset Loader (DFR Dataset) - Modified for EfficientNet-B3
import os
import cv2
import torch
from torch.utils.data import Dataset
from torchvision import transforms

class DFRDataset(Dataset):
    def __init__(self, root_dir):
        self.root_dir = root_dir
        self.classes = os.listdir(root_dir)
        self.image_paths = []
        self.labels = []

        for label in self.classes:
            class_path = os.path.join(root_dir, label)
            for img in os.listdir(class_path):
                self.image_paths.append(os.path.join(class_path, img))
                self.labels.append(int(label))

        self.weather = get_weather_transform()

        # EfficientNet-B3 was trained with ImageNet normalization
        # EfficientNet-B3 uses 300x300 input size (different from ResNet's 224)
        self.transform = transforms.Compose([
            transforms.ToTensor(),
            transforms.Resize((300, 300)),  # EfficientNet-B3 expects 300x300
            transforms.Normalize(mean=[0.485, 0.456, 0.406],
                               std=[0.229, 0.224, 0.225])
        ])

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img = cv2.imread(self.image_paths[idx])
        if img is None:
            print(f"Warning: Could not read image {self.image_paths[idx]}")
            return self.__getitem__((idx + 1) % len(self.image_paths))

        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        # Weather simulation
        img = self.weather(image=img)['image']

        # Enhancement
        img = clahe_enhancement(img)
        img = sharpen_image(img)

        img = self.transform(img)

        label = torch.tensor(self.labels[idx])

        return img, label
```

```python
# EfficientNet-B3 Model for Traffic Sign Detection
import torch
import torch.nn as nn
import torchvision.models as models

class EfficientNetB3_TSD(nn.Module):
    def __init__(self, num_classes):
        super(EfficientNetB3_TSD, self).__init__()

        # Load pretrained EfficientNet-B3
        # Using weights='DEFAULT' instead of deprecated pretrained=True
        self.backbone = models.efficientnet_b3(weights='DEFAULT')

        # EfficientNet-B3 architecture:
        # - features: convolutional backbone
        # - classifier: Sequential(
        #     Dropout(p=0.3),
        #     Linear(1536, 1000)  # 1536 is the feature dimension for B3
        #   )

        # Get the feature dimension (1536 for efficientnet_b3)
        self.feature_dim = 1536

        # Remove the original classifier
        # EfficientNet's classifier is a Sequential with Dropout + Linear
        # We'll replace the entire classifier
        self.backbone.classifier = nn.Identity()

        # Custom Classifier Head (similar to your ResNet50 design)
        # EfficientNet-B3 already has built-in dropout, but we'll add our own
        self.classifier = nn.Sequential(
            nn.Dropout(0.3),  # EfficientNet's default dropout
            nn.Linear(self.feature_dim, 512),
            nn.ReLU(),
            nn.Dropout(0.4),
            nn.Linear(512, num_classes)
        )

    def forward(self, x):
        # Extract features using EfficientNet-B3 backbone
        x = self.backbone(x)  # Shape: [batch_size, 1536]

        # Classify using custom head
        x = self.classifier(x)  # Shape: [batch_size, num_classes]

        return x
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, random_split
from tqdm import tqdm
import matplotlib.pyplot as plt
import os
import time

# Set device
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {DEVICE}")

# Load Dataset
dataset_path = "/content/drive/MyDrive/SEM-8 MAJOR PROJECT/Images"
if not os.path.exists(dataset_path):
    print(f"Error: Dataset path {dataset_path} does not exist!")

dataset = DFRDataset(dataset_path)
print(f"Total images in dataset: {len(dataset)}")

# Split into Train & Validation (80-20)
train_size = int(0.8 * len(dataset))
val_size = len(dataset) - train_size
train_dataset, val_dataset = random_split(dataset, [train_size, val_size])

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=2, pin_memory=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=2, pin_memory=True)

num_classes = 59  # Number of traffic sign classes

# Initialize EfficientNet-B3 model
model = EfficientNetB3_TSD(num_classes).to(DEVICE)

# Count parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Total parameters: {total_params:,}")
print(f"Trainable parameters: {trainable_params:,}")

# Loss function and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.AdamW(model.parameters(), lr=0.0003)  # Same LR as ResNet50 success

# Learning rate scheduler - StepLR works well with EfficientNet
scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.1)

# Optional: Cosine Annealing scheduler (often better for EfficientNet)
# scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=10)

EPOCHS = 10  # Same as your previous successful runs

# Store history
train_losses = []
val_losses = []
train_accuracies = []
val_accuracies = []
best_val_acc = 0

print("Starting training with EfficientNet-B3...")
start_time = time.time()

for epoch in range(EPOCHS):
    # Training phase
    model.train()
    train_loss = 0
    correct = 0
    total = 0

    train_bar = tqdm(train_loader, desc=f'Epoch {epoch+1}/{EPOCHS} [Train]')
    for images, labels in train_bar:
        images, labels = images.to(DEVICE), labels.to(DEVICE)

        # Forward pass
        outputs = model(images)
        loss = criterion(outputs, labels)

        # Backward pass and optimize
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        train_loss += loss.item()

        # Calculate accuracy
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

        # Update progress bar
        train_bar.set_postfix({'loss': loss.item(), 'acc': 100 * correct/total})

    train_accuracy = 100 * correct / total
    avg_train_loss = train_loss / len(train_loader)

    # Validation phase
    model.eval()
    val_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():
        val_bar = tqdm(val_loader, desc=f'Epoch {epoch+1}/{EPOCHS} [Val]')
        for images, labels in val_bar:
            images, labels = images.to(DEVICE), labels.to(DEVICE)
            outputs = model(images)
            loss = criterion(outputs, labels)

            val_loss += loss.item()

            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

            val_bar.set_postfix({'loss': loss.item(), 'acc': 100 * correct/total})

    val_accuracy = 100 * correct / total
    avg_val_loss = val_loss / len(val_loader)

    # Update learning rate
    scheduler.step()

    # Store metrics
    train_losses.append(avg_train_loss)
    val_losses.append(avg_val_loss)
    train_accuracies.append(train_accuracy)
    val_accuracies.append(val_accuracy)

    # Save best model
    if val_accuracy > best_val_acc:
        best_val_acc = val_accuracy
        best_model_path = "/content/drive/MyDrive/SEM-8 MAJOR PROJECT/models/EfficientNetB3_TSD_best.pth"
        os.makedirs(os.path.dirname(best_model_path), exist_ok=True)
        torch.save(model.state_dict(), best_model_path)
        print(f" Saved best model with validation accuracy: {val_accuracy:.2f}% (Epoch {epoch+1})")

    print(f"\nEpoch [{epoch+1}/{EPOCHS}]")
    print(f"Train Loss: {avg_train_loss:.4f}, Train Acc: {train_accuracy:.2f}%")
    print(f"Val Loss: {avg_val_loss:.4f}, Val Acc: {val_accuracy:.2f}%")
    print(f"Current LR: {scheduler.get_last_lr()[0]:.6f}")
    print("-" * 60)

# Save final model
final_model_path = "/content/drive/MyDrive/SEM-8 MAJOR PROJECT/models/EfficientNetB3_TSD_final.pth"
torch.save(model.state_dict(), final_model_path)

end_time = time.time()
training_time = end_time - start_time
print(f"\n Training completed in {training_time//60:.0f}m {training_time%60:.0f}s")
print(f"Best validation accuracy: {best_val_acc:.2f}%")
```
